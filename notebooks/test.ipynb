```python
from vantara_customer_intelligence_platform.utils.types import PandasFeaturesSchema
from vantara_customer_intelligence_platform.utils.data import generate_schema
from pathlib import Path
import pandas as pd
```

```python
# Extract the regular types and date columns dynamically from your schema dictionary
regular_dtypes = {col: dt for col, dt in PandasFeaturesSchema.items() if not str(dt).startswith('datetime')}
date_columns = [col for col, dt in PandasFeaturesSchema.items() if str(dt).startswith('datetime')]
```

```python
df = pd.read_csv("data/raw/data.csv", dtype=regular_dtypes, parse_dates=date_columns)

data = pd.read_excel(Path("data/raw/online_retail_II.xlsx").absolute(), sheet_name=None, dtype=regular_dtypes, parse_dates=date_columns)
df_org = pd.concat(data.values(), ignore_index=True)
```
